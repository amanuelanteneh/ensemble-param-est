```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import halfnorm, lognorm, foldnorm
```

```python
# Parameters
loc = 1      # mean of the distribution (must be ≥ 0 for half-normal)
scale = 1    # standard deviation

# x range (starting from 0, since half-normal is only defined for x ≥ 0)
x = np.linspace(0, 5, 1000)

# PDF
pdf = halfnorm.pdf(x, loc=loc, scale=scale)

# Plot
plt.plot(x, pdf, label=f'Half-Normal PDF\n(loc={loc}, scale={scale})')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.title('Half-Normal Distribution PDF')
plt.legend();
```

```python
# Parameters
c = 1.5 
loc = 0.5      # mean of the distribution (must be ≥ 0 for half-normal)
scale = 1    # standard deviation

# x range (starting from 0, since half-normal is only defined for x ≥ 0)
x = np.linspace(0, 5, 1000)

# PDF
pdf = foldnorm.pdf(x, loc=loc, scale=scale, c=c)
print(foldnorm.mean(c, loc=loc, scale=scale))

# Plot
plt.plot(x, pdf, label=f'Folded Normal PDF\n(loc={loc}, scale={scale})')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.title('Folded-Normal Distribution PDF')
plt.legend();
```

```python
# Parameters
sigma = 0.9             # standard deviation of the underlying normal distribution
scale = np.exp(.1)     # exp(mu), where mu is the mean of the underlying normal
loc = 2                 # usually 0

# x range (log-normal is defined for x > 0)
x = np.linspace(0.001, 5, 1000)

# PDF
pdf = lognorm.pdf(x, s=sigma, loc=loc, scale=scale)

# Plot
plt.plot(x, pdf, label=f'Log-Normal PDF\n(sigma={sigma}, scale={scale})')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.title('Log-Normal Distribution PDF')
plt.legend();
```
